# Tutorial: Analyzing the Geometry of a 2D Sphere with `gr`

This notebook explores a classic example from differential geometry: the surface of a two-dimensional sphere embedded in three-dimensional Euclidean space. We will use the `gr` Python module, typically used for General Relativity, to calculate the geometric tensors (Christoffel symbols, Riemann tensor, etc.) for the sphere's intrinsic geometry.

## 1. Introduction: Geometry of a 2D Sphere

Imagine a sphere of constant radius $R$ centered at the origin in 3D Euclidean space ($x^2+y^2+z^2=R^2$). We can describe points on the surface of this sphere using spherical coordinates $(\theta, \phi)$, where:
* $\theta$ is the polar angle (angle from the z-axis, $0 \le \theta \le \pi$).
* $\phi$ is the azimuthal angle (angle in the x-y plane, $0 \le \phi < 2\pi$).
* The radius $r$ is fixed at $R$.

The line element $ds^2$ represents the infinitesimal squared distance between two nearby points on the surface. In the 3D Euclidean space, $ds^2 = dx^2 + dy^2 + dz^2$. By substituting $x=R\sin\theta\cos\phi$, $y=R\sin\theta\sin\phi$, $z=R\cos\theta$ and calculating $dx, dy, dz$, we find the line element restricted *to the surface* of the sphere is:

$ds^2 = R^2 d\theta^2 + R^2 \sin^2\theta d\phi^2$

This describes the intrinsic geometry of the sphere – the geometry experienced by a hypothetical 2D being living only on the surface, unaware of the embedding 3D space.

From the line element, we can read off the components of the 2D metric tensor $g_{ij}$ in coordinates $(\theta, \phi)$:
* $g_{\theta\theta} = R^2$
* $g_{\phi\phi} = R^2 \sin^2\theta$
* $g_{\theta\phi} = g_{\phi\theta} = 0$

**Key Geometric Properties:**
* **Intrinsic Curvature:** A sphere is intrinsically curved. This means its geometry cannot be "flattened" onto a plane without stretching or tearing (unlike a cylinder, which can be unrolled into a flat rectangle). This curvature should be reflected in a non-zero Riemann tensor.
* **Constant Curvature:** The sphere has constant positive Gaussian curvature $K = 1/R^2$. In 2D, the Ricci scalar $R$ is related to the Gaussian curvature by $R = 2K$. Thus, we expect the Ricci scalar to be $R = 2/R^2$.
* **Ricci Tensor:** For a 2D manifold of constant curvature, the Ricci tensor is proportional to the metric tensor: $R_{ij} = K g_{ij} = \frac{R}{2} g_{ij}$. We expect $R_{ij} = \frac{1}{R^2} g_{ij}$.
* **Einstein Tensor (2D):** The Einstein tensor is defined as $G_{ij} = R_{ij} - \frac{1}{2} g_{ij} R$. Substituting the expected relationship $R_{ij} = \frac{R}{2} g_{ij}$ for 2D, we get $G_{ij} = \frac{R}{2} g_{ij} - \frac{1}{2} g_{ij} R = 0$. The Einstein tensor is identically zero in 2 dimensions.

Let's use the `gr` module to compute these quantities and verify these properties.

```python
# --- Define Coordinates, Parameters, and Metric ---

# Define symbols for 2D spherical coordinates (theta, phi)
# Assume theta is in (0, pi) to avoid issues with sin(theta)=0 at poles if needed later.
# Assume phi is in (0, 2pi).
theta, phi = symbols("θ φ", real=True, positive=True)
coords_2d = [theta, phi]
print("Coordinates:")
display(coords_2d)

# Define the constant radius R (positive real number)
R = symbols("R", real=True, positive=True)
print("\nRadius Parameter:")
display(R)

# Define the metric tensor components for the 2D sphere
g_thetatheta = R**2
g_phiphi = R**2 * sin(theta)**2

# Construct the 2x2 metric tensor as a SymPy Matrix
metric_sphere_2d = Matrix([
    [ g_thetatheta,        0       ],
    [      0      ,   g_phiphi   ]
])

print("\n2D Sphere Metric Tensor g_ij:")
# Display the metric using LaTeX
display(Math(latex(metric_sphere_2d)))

# Display its components for clarity
print("\nMetric Components:")
display(Math(f"g_{{θθ}} = {latex(g_thetatheta)}"))
display(Math(f"g_{{φφ}} = {latex(g_phiphi)}"))
```

## 2. Christoffel Symbols ($Γ^k_{ij}$)

The Christoffel symbols for the sphere describe how basis vectors change as one moves across the curved surface in the $(\theta, \phi)$ coordinate system.

$\Gamma^k_{ij} = \frac{1}{2} g^{kl} \left( \frac{\partial g_{jl}}{\partial x^i} + \frac{\partial g_{il}}{\partial x^j} - \frac{\partial g_{ij}}{\partial x^l} \right)$
(Indices $i, j, k, l$ run over $\theta, \phi$)

```python
# --- Compute and Display Christoffel Symbols ---

print("Computing Christoffel Symbols Γ^k_{ij}...")
# Use the 2D coordinates and 2x2 metric
christoffel_symbols_2d = compute_christoffel_symbols(coords_2d, metric_sphere_2d)

# Display the non-zero Christoffel symbols
if christoffel_symbols_2d:
    display_christoffel_symbols(christoffel_symbols_2d, coords_2d)
else:
    print("Could not compute Christoffel symbols (check 'gr' module import).")
```

## 3. Riemann Curvature Tensor ($R^k_{l i j}$)

The Riemann tensor quantifies the intrinsic curvature of the sphere's surface. Since the sphere is curved, we expect non-zero components. In 2D, the Riemann tensor has only one independent component (up to symmetries).

$R^k_{l i j} = \frac{\partial \Gamma^k_{j l}}{\partial x^i} - \frac{\partial \Gamma^k_{i l}}{\partial x^j} + \Gamma^k_{i m} \Gamma^m_{j l} - \Gamma^k_{j m} \Gamma^m_{i l}$
(Indices run over $\theta, \phi$)

```python
# --- Compute and Display Riemann Curvature Tensor ---

print("Computing Riemann Curvature Tensor R^k_{l i j}...")
# Pass the pre-computed Christoffel symbols
riemann_tensor_2d = compute_riemann_curvature_tensor(coords_2d, metric_sphere_2d, christoffel_symbols=christoffel_symbols_2d)

# Display the non-zero components
# Expect non-zero components indicating curvature
if riemann_tensor_2d:
    display_riemann_curvature_tensor(riemann_tensor_2d, coords_2d)
else:
    print("Could not compute Riemann tensor.")
```

## 4. Ricci Curvature Tensor ($R_{ij}$)

The Ricci tensor is the contraction $R_{ij} = R^k_{i k j}$. For the 2D sphere, due to its constant curvature $K=1/R^2$, we expect the Ricci tensor to be proportional to the metric: $R_{ij} = K g_{ij} = \frac{1}{R^2} g_{ij}$.

```python
# --- Compute and Display Ricci Tensor ---

print("Computing Ricci Tensor R_{ij}...")
# Pass the pre-computed Riemann tensor
ricci_tensor_2d = compute_ricci_tensor(coords_2d, metric_sphere_2d, riemann_tensor=riemann_tensor_2d)

# Display the components
# Expect R_θθ = 1, R_φφ = sin^2(θ), R_θφ = 0 (after dividing by R^2 from metric)
# i.e., R_ij = (1/R^2) * g_ij
if ricci_tensor_2d:
    print("Displaying Ricci Tensor Components:")
    display_ricci_tensor(ricci_tensor_2d, coords_2d)

    # Verify the relationship R_ij = (1/R^2) * g_ij
    expected_ricci = (1/R**2) * metric_sphere_2d
    print("\nVerifying if R_ij = (1/R^2) * g_ij:")
    diff_matrix = sympy.simplify(Matrix(ricci_tensor_2d) - expected_ricci)
    is_relation_valid = (diff_matrix == sympy.zeros(2, 2))

    if is_relation_valid:
        print("Verification successful: R_ij = (1/R^2) * g_ij holds.")
        display(Math(f"R_{{ij}} = \\frac{{1}}{{R^2}} g_{{ij}} = {latex(expected_ricci)}"))
    else:
        print("Warning: Relationship R_ij = (1/R^2) * g_ij did not verify!")
        print("Computed Ricci:")
        display(Matrix(ricci_tensor_2d))
        print("Expected Ricci:")
        display(expected_ricci)
else:
    print("Could not compute Ricci tensor.")
```

## 5. Ricci Scalar ($R$)

The Ricci scalar is $R = g^{ij} R_{ij}$. For a 2D sphere, it should be constant and equal to $2K = 2/R^2$.

```python
# --- Compute and Display Ricci Scalar ---

print("Computing Ricci Scalar R...")
# Pass the pre-computed Ricci tensor
ricci_scalar_2d = compute_ricci_scalar(coords_2d, metric_sphere_2d, ricci_tensor=ricci_tensor_2d)

# Display the result (expecting 2/R^2)
expected_scalar = 2/R**2
if ricci_scalar_2d is not None:
    print("Displaying Ricci Scalar:")
    display_ricci_scalar(ricci_scalar_2d)
    # Explicit check
    if simplify(ricci_scalar_2d - expected_scalar) == 0:
        print(f"\nVerification successful: The Ricci scalar R = 2/R^2.")
    else:
        print(f"\nWarning: The Ricci scalar did not simplify to 2/R^2! Result: {ricci_scalar_2d}")
else:
    print("Could not compute Ricci scalar.")
```

## 6. Einstein Tensor ($G_{ij}$)

The Einstein tensor is $G_{ij} = R_{ij} - \frac{1}{2} g_{ij} R$. As noted earlier, for any 2-dimensional manifold, the relationship $R_{ij} = \frac{R}{2} g_{ij}$ holds (this is specific to 2D). Therefore, the Einstein tensor in 2D should always be identically zero.

```python
# --- Compute and Display Einstein Tensor ---

print("Computing Einstein Tensor G_{ij}...")
# Pass the pre-computed Ricci tensor and scalar
einstein_tensor_2d = compute_einstein_tensor(coords_2d, metric_sphere_2d, ricci_tensor=ricci_tensor_2d, ricci_scalar=ricci_scalar_2d)

# Display the components (should all be zero)
if einstein_tensor_2d:
    print("Displaying Einstein Tensor Components (expecting zero):")
    display_einstein_tensor(einstein_tensor_2d, coords_2d)
    # Explicit check
    is_zero = all(simplify(comp) == 0 for row in einstein_tensor_2d for comp in row)
    if is_zero:
        print("\nVerification successful: All components of the Einstein tensor are zero.")
    else:
        print("\nWarning: Some components of the Einstein tensor did not simplify to zero!")
        # Optionally display the matrix form for inspection
        display(Matrix(einstein_tensor_2d))
else:
    print("Could not compute Einstein tensor.")
```

## 7. Conclusion

We have analyzed the intrinsic geometry of a 2D sphere using the `gr` module. The key findings are:

* The **Christoffel symbols** are non-zero, reflecting the curvilinear nature of the $(\theta, \phi)$ coordinates.
* The **Riemann tensor** is non-zero, confirming that the sphere is an intrinsically curved surface.
* The **Ricci tensor** is proportional to the metric tensor ($R_{ij} = \frac{1}{R^2} g_{ij}$), characteristic of a space with constant curvature.
* The **Ricci scalar** is constant and positive ($R = 2/R^2$), directly related to the sphere's Gaussian curvature ($K=1/R^2$).
* The **Einstein tensor** is identically zero, a general property of 2-dimensional geometry/gravity.

These results match the well-known geometric properties of a sphere and demonstrate that the `gr` module can be effectively used for analyzing the geometry of manifolds in dimensions other than four.
